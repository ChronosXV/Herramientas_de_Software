````python
# Ejercicio 2.1

# Listar todas las imágenes disponibles

image_dir: str = os.path.join(output_dir, 'urban_flow_plates')

# Lista para almacenar la información de las imágenes
image_info_list: list[dict[str, str | float]] = []

print("Listado de imágenes y su tamaño en KB:")
print("------------------------------------")

# Recorrer el directorio para listar las imágenes y sus tamaños
if os.path.exists(image_dir):
    for filename in os.listdir(image_dir):
        file_path = os.path.join(image_dir, filename)
        if os.path.isfile(file_path):
            # Obtener el tamaño del archivo en bytes
            size_bytes = os.path.getsize(file_path)
            # Convertir a kilobytes
            size_kb = size_bytes / 1024
            image_info_list.append({"filename": filename, "size_kb": size_kb})
            print(f"Nombre: {filename}, Tamaño: {size_kb:.2f} KB")
else:
    print(f"El directorio de imágenes no existe: {image_dir}")

print("------------------------------------")
# La lista 'image_info_list' contiene la información de las imágenes
# para ser utilizada en pasos posteriores sin necesidad de volver a listar.
````

````python
# Ejercicio 2.2

# Listar todas las imágenes disponibles y separarlas en grupos 'plates' y 'completes'

image_dir: str = os.path.join(output_dir, 'urban_flow_plates')

# Diccionario para almacenar la información de las imágenes agrupadas
group_images: dict[str, list[dict[str, str | int | float]]] = {
    'plates': [],
    'completes': [],
}

# Recorrer el directorio para listar y clasificar las imágenes
if os.path.exists(image_dir):
    for filename in os.listdir(image_dir):
        file_path = os.path.join(image_dir, filename)
        if os.path.isfile(file_path):
            try:
                # Abrir la imagen para obtener sus dimensiones
                with Image.open(file_path) as img:
                    width, height = img.size
                    area = width * height

                image_info: dict[str, str | int | float] = {
                    'filename': filename,
                    'width': width,
                    'height': height,
                    'area': area,
                    'path': file_path,
                    'patent': '',  # Inicializar la patente como vacía
                }

                # Clasificar la imagen en 'plates' o 'completes'
                if 'patente' in filename.lower():
                    group_images['plates'].append(image_info)
                else:
                    group_images['completes'].append(image_info)

            except Image.UnidentifiedImageError:
                print(
                    f"Advertencia: No se pudo identificar el formato de imagen para {filename}. Se omitirá."
                )
            except Exception as e:
                print(
                    f"Error al procesar la imagen {filename}: {e}. Se omitirá."
                )

    # Imprimir la cantidad de imágenes en cada grupo
    print(f"El grupo 'plates' contiene {len(group_images['plates'])} imágenes.")
    print(f"El grupo 'completes' contiene {len(group_images['completes'])} imágenes.")

    # Guardar el diccionario group_images en un archivo JSON
    output_json_path = 'urban_flow/data/interim/group_images.json'
    os.makedirs(os.path.dirname(output_json_path), exist_ok=True)
    with open(output_json_path, 'w') as f:
        json.dump(group_images, f, indent=2)
    print(f"Información de las imágenes guardada en {output_json_path}")

else:
    print(f"El directorio de imágenes no existe: {image_dir}")
````

````python
# Ejercicio 2.3

# Desarrollar un función para mostrar 8 imágenes de forma aleatorea

def display_random_images(group_images: dict, num_images_per_group: int = 4) -> None:
    """
    Muestra imágenes aleatorias de los grupos 'plates' y 'completes' en una cuadrícula.

    Args:
        group_images (dict): Diccionario que contiene la información de las imágenes agrupadas.
                             Debe tener las claves 'plates' y 'completes', cada una con una lista de diccionarios
                             que incluyen 'path' y 'filename'.
        num_images_per_group (int): Número de imágenes a mostrar por cada grupo.
    """
    selected_plates = random.sample(
        group_images['plates'], min(num_images_per_group, len(group_images['plates']))
    )
    selected_completes = random.sample(
        group_images['completes'],
        min(num_images_per_group, len(group_images['completes'])),
    )

    # Combinar las imágenes seleccionadas
    all_selected_images = selected_plates + selected_completes
    random.shuffle(all_selected_images)  # Mezclar para una visualización aleatoria

    # Asegurar que no se excedan las 8 imágenes si los grupos son pequeños
    display_count = min(len(all_selected_images), 8)
    images_to_display = all_selected_images[:display_count]

    if not images_to_display:
        print("No hay imágenes disponibles para mostrar.")
        return

    # Configurar la figura y los subplots
    # 4 filas, 2 columnas para un total de 8 imágenes
    fig, axes = plt.subplots(
        nrows=4, ncols=2, figsize=(2 * 10, 4 * 8)
    )  # 10 pulgadas de ancho, 8 pulgadas de alto por imagen
    axes = axes.flatten()

    for i, img_info in enumerate(images_to_display):
        if i >= 8:  # Asegurar que no excedemos los 8 subplots
            break
        try:
            img = Image.open(img_info['path'])
            axes[i].imshow(img)
            axes[i].set_title(img_info['filename'], fontsize=10)
            axes[i].axis('off')  # Ocultar ejes
        except FileNotFoundError:
            axes[i].set_title(
                f"Error: {img_info['filename']} no encontrada",
                fontsize=10,
                color='red',
            )
            axes[i].axis('off')
        except Exception as e:
            axes[i].set_title(
                f"Error al cargar {img_info['filename']}: {e}",
                fontsize=10,
                color='red',
            )
            axes[i].axis('off')

    # Ocultar subplots vacíos si hay menos de 8 imágenes
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout()
    plt.show()


# Llamar a la función para mostrar 8 imágenes (4 de cada grupo)
display_random_images(group_images)
````

````python
# Ejercicio 3.1

# Creación de directorios de salida para las imágenes en escala de grises
base_interim_img_dir: str = "urban_flow/data/interim/imgs"
gray_scale_dir_name: str = "03_01_gray_scale"
output_gray_scale_plates_dir: str = os.path.join(
    base_interim_img_dir, gray_scale_dir_name, "plates"
)
output_gray_scale_completes_dir: str = os.path.join(
    base_interim_img_dir, gray_scale_dir_name, "completes"
)

os.makedirs(output_gray_scale_plates_dir, exist_ok=True)
os.makedirs(output_gray_scale_completes_dir, exist_ok=True)


def convert_image_to_grayscale_and_save(input_path: str,
                                      output_path: str) -> str:
    """
    Convierte una imagen a escala de grises y la guarda en la ruta especificada.

    Args:
        input_path (str): Ruta completa de la imagen de entrada.
        output_path (str): Ruta completa donde se guardará la imagen en
                           escala de grises.

    Returns:
        str: La ruta del archivo guardado en escala de grises, o una cadena
             vacía si falla.
    """
    try:
        with Image.open(input_path) as img:
            gray_img = img.convert('L')  # Convertir a escala de grises
            gray_img.save(output_path)
            return output_path
    except FileNotFoundError:
        print(f"Error: La imagen no se encontró en {input_path}")
        return ""
    except Exception as e:
        print(f"Error al procesar la imagen {input_path}: {e}")
        return ""


# Diccionario para almacenar la información de las imágenes en escala de grises
grayscale_group_images: dict[str, list[dict[str, str | int | float]]] = {
    'plates': [],
    'completes': [],
}

# Procesar imágenes del grupo 'plates'
for img_info in group_images['plates']:
    original_path: str = img_info['path']
    filename: str = img_info['filename']
    output_path: str = os.path.join(output_gray_scale_plates_dir, filename)

    saved_path: str = convert_image_to_grayscale_and_save(
        original_path, output_path
    )

    if saved_path:
        # Crear una copia de img_info y actualizar la ruta para las imágenes
        # en escala de grises
        gray_img_info = img_info.copy()
        gray_img_info['path'] = saved_path
        grayscale_group_images['plates'].append(gray_img_info)

# Procesar imágenes del grupo 'completes'
for img_info in group_images['completes']:
    original_path: str = img_info['path']
    filename: str = img_info['filename']
    output_path: str = os.path.join(
        output_gray_scale_completes_dir, filename
    )

    saved_path: str = convert_image_to_grayscale_and_save(
        original_path, output_path
    )

    if saved_path:
        # Crear una copia de img_info y actualizar la ruta para las imágenes
        # en escala de grises
        gray_img_info = img_info.copy()
        gray_img_info['path'] = saved_path
        grayscale_group_images['completes'].append(gray_img_info)

print("Imágenes convertidas a escala de grises y guardadas.")

# Mostrar una selección de las imágenes en escala de grises utilizando la
# función del Ejercicio 2.3
print("\nMostrando una selección de imágenes en escala de grises:")
display_random_images(grayscale_group_images)

# Almacenar el estado actual de las imágenes procesadas para el siguiente paso
current_processed_images = grayscale_group_images
````

````python
# Ejercicio 3.2

# Suavizado de las imágenes en escala de grises

# 1. Crear directorios de salida para las imágenes suavizadas
base_interim_img_dir: str = "urban_flow/data/interim/imgs"
blur_dir_name: str = "03_02_blur"
output_blur_plates_dir: str = os.path.join(
    base_interim_img_dir, blur_dir_name, "plates"
)
output_blur_completes_dir: str = os.path.join(
    base_interim_img_dir, blur_dir_name, "completes"
)

os.makedirs(output_blur_plates_dir, exist_ok=True)
os.makedirs(output_blur_completes_dir, exist_ok=True)


def apply_gaussian_blur_and_save(
    input_path: str, output_path: str, radius: float = 2.0
) -> str:
    """
    Aplica un filtro Gaussiano a una imagen y la guarda en la ruta especificada.

    Args:
        input_path (str): Ruta completa de la imagen de entrada (debe ser en
                          escala de grises).
        output_path (str): Ruta completa donde se guardará la imagen suavizada.
        radius (float): El radio del filtro Gaussiano.

    Returns:
        str: La ruta del archivo guardado, o una cadena vacía si falla.
    """
    try:
        with Image.open(input_path) as img:
            # Asegurarse de que la imagen esté en modo 'L' (escala de grises)
            if img.mode != 'L':
                img = img.convert('L')
            blurred_img = img.filter(ImageFilter.GaussianBlur(radius))
            blurred_img.save(output_path)
            return output_path
    except FileNotFoundError:
        print(f"Error: La imagen no se encontró en {input_path}")
        return ""
    except Exception as e:
        print(f"Error al procesar la imagen {input_path}: {e}")
        return ""


# Diccionario para almacenar la información de las imágenes suavizadas
blurred_group_images: dict[str, list[dict[str, str | int | float]]] = {
    'plates': [],
    'completes': [],
}

# Procesar imágenes del grupo 'plates' (que ahora son las de escala de grises)
for img_info in current_processed_images['plates']:
    original_path: str = img_info['path']
    filename: str = img_info['filename']
    output_path: str = os.path.join(output_blur_plates_dir, filename)

    saved_path: str = apply_gaussian_blur_and_save(original_path, output_path)

    if saved_path:
        # Crear una copia de img_info y actualizar la ruta para las imágenes
        # suavizadas
        blur_img_info = img_info.copy()
        blur_img_info['path'] = saved_path
        blurred_group_images['plates'].append(blur_img_info)

# Procesar imágenes del grupo 'completes' (que ahora son las de escala de grises)
for img_info in current_processed_images['completes']:
    original_path: str = img_info['path']
    filename: str = img_info['filename']
    output_path: str = os.path.join(output_blur_completes_dir, filename)

    saved_path: str = apply_gaussian_blur_and_save(original_path, output_path)

    if saved_path:
        # Crear una copia de img_info y actualizar la ruta para las imágenes
        # suavizadas
        blur_img_info = img_info.copy()
        blur_img_info['path'] = saved_path
        blurred_group_images['completes'].append(blur_img_info)

print("Imágenes en escala de grises suavizadas y guardadas.")

# Mostrar una selección de las imágenes suavizadas
print("\nMostrando una selección de imágenes suavizadas:")
display_random_images(blurred_group_images)

# Almacenar el estado actual de las imágenes procesadas para el siguiente paso
current_processed_images = blurred_group_images
````

````python
# Ejercicio 3.3

# Detección de bordes sobre las imágenes suavizadas

# Creación directorios de salida para las imágenes con detección de bordes Canny
base_interim_img_dir: str = "urban_flow/data/interim/imgs"
canny_dir_name: str = "03_03_canny"
output_canny_plates_dir: str = os.path.join(
    base_interim_img_dir, canny_dir_name, "plates"
)
output_canny_completes_dir: str = os.path.join(
    base_interim_img_dir, canny_dir_name, "completes"
)

os.makedirs(output_canny_plates_dir, exist_ok=True)
os.makedirs(output_canny_completes_dir, exist_ok=True)


def apply_canny_and_save(
    input_path: str,
    output_path: str,
    low_threshold: int = 50,
    high_threshold: int = 150,
) -> str:
    """
    Aplica el algoritmo Canny para detección de bordes y guarda la imagen.

    Args:
        input_path (str): Ruta completa de la imagen de entrada (debe ser en
                          escala de grises).
        output_path (str): Ruta completa donde se guardará la imagen con
                           bordes detectados.
        low_threshold (int): Primer umbral para el procedimiento de histéresis.
        high_threshold (int): Segundo umbral para el procedimiento de
                              histéresis.

    Returns:
        str: La ruta del archivo guardado, o una cadena vacía si falla.
    """
    try:
        # Cargar la imagen usando PIL y convertir a NumPy array para OpenCV
        with Image.open(input_path) as img_pil:
            # Asegurarse de que la imagen está en escala de grises para Canny
            if img_pil.mode != 'L':
                img_pil = img_pil.convert('L')
            img_np = np.array(img_pil)

        # Aplicar el algoritmo Canny
        edges = cv2.Canny(img_np, low_threshold, high_threshold)

        # Guardar la imagen resultante (convertir de NumPy array a PIL Image
        # para guardar)
        Image.fromarray(edges).save(output_path)
        return output_path
    except FileNotFoundError:
        print(f"Error: La imagen no se encontró en {input_path}")
        return ""
    except Exception as e:
        print(f"Error al procesar la imagen {input_path}: {e}")
        return ""


# Diccionario para almacenar la información de las imágenes con bordes Canny
canny_group_images: dict[str, list[dict[str, str | int | float]]] = {
    'plates': [],
    'completes': [],
}

# Procesar imágenes del grupo 'plates' (que ahora son las imágenes suavizadas)
for img_info in current_processed_images['plates']:
    original_path: str = img_info['path']
    filename: str = img_info['filename']
    output_path: str = os.path.join(output_canny_plates_dir, filename)

    saved_path: str = apply_canny_and_save(original_path, output_path)

    if saved_path:
        canny_img_info = img_info.copy()
        canny_img_info['path'] = saved_path
        canny_group_images['plates'].append(canny_img_info)

# Procesar imágenes del grupo 'completes' (que ahora son las imágenes suavizadas)
for img_info in current_processed_images['completes']:
    original_path: str = img_info['path']
    filename: str = img_info['filename']
    output_path: str = os.path.join(output_canny_completes_dir, filename)

    saved_path: str = apply_canny_and_save(original_path, output_path)

    if saved_path:
        canny_img_info = img_info.copy()
        canny_img_info['path'] = saved_path
        canny_group_images['completes'].append(canny_img_info)

print("Imágenes con detección de bordes Canny guardadas.")

# Mostrar una selección de las imágenes con detección de bordes Canny
print("\nMostrando una selección de imágenes con detección de bordes Canny:")
display_random_images(canny_group_images)

# Almacenar el estado actual de las imágenes procesadas para el siguiente paso
current_processed_images = canny_group_images
````

````python
# Definición de la función extraer_patente

import easyocr
def extraer_patente(image_path) -> str:
  """ Extrae la patente de una imagen.

  Args:
      image_path (str): Ruta de la imagen.

  Returns:
      str: Patente extraída.
  """
  reader = easyocr.Reader(['es'])
  detection:list = reader.readtext(image_path)
  if len(detection) == 0:
    return ''
  patent = ''
  for d in detection:
    if isinstance(d,tuple) and len(d) == 3:
      if isinstance(d[1],str):
        patent+=d[1]
  return patent
````

````python
# Ejercicio 4

# Función para calcular el ratio de coincidencia entre patentes
def calculate_match_ratio(
    original_patent: str,
    detected_patent: str
) -> float:
    """
    Calcula el porcentaje de caracteres coincidentes en la misma posición
    entre dos patentes.

    Args:
        original_patent (str): Patente original del dataset.
        detected_patent (str): Patente detectada mediante OCR.

    Returns:
        float: Ratio de coincidencia entre 0.0 y 1.0.
    """

    # Evitar comparación con cadenas vacías
    if not detected_patent:
        return 0.0

    original_upper = original_patent.upper()
    detected_upper = detected_patent.upper()

    matches = 0

    # Comparar caracteres de izquierda a derecha
    for i in range(
        min(len(original_upper), len(detected_upper))
    ):

        if original_upper[i] == detected_upper[i]:
            matches += 1

    # Calcular ratio utilizando la longitud máxima
    return matches / max(
        len(original_upper),
        len(detected_upper)
    )


# Lista para almacenar información de patentes detectadas
detected_patents_data: list[dict[str, str]] = []

print("Extrayendo patentes de las imágenes...")


# OCR se realiza sobre imágenes en escala de grises
for group_key in grayscale_group_images:

    for img_info in grayscale_group_images[group_key]:

        # Si aún no se extrajo la patente
        if not img_info.get('patent'):

            try:

                patent_ocr = extraer_patente(
                    img_info['path']
                )

                img_info['patent'] = patent_ocr

                # Guardar solo si se detectó texto
                if patent_ocr:

                    detected_patents_data.append({
                        'path': img_info['path'],
                        'patent_ocr': patent_ocr
                    })

            except Exception as e:

                print(
                    f"Error al extraer patente "
                    f"de {img_info['path']}: {e}"
                )

                img_info['patent'] = "ERROR_OCR"

        # Si ya existe una patente válida
        elif (
            img_info['patent'] != "ERROR_OCR"
            and img_info['patent']
        ):

            detected_patents_data.append({
                'path': img_info['path'],
                'patent_ocr': img_info['patent']
            })


print("Extracción de patentes completada.")


# Mostrar algunas patentes detectadas para verificación
print("\nEjemplos de patentes detectadas:")

for patent_info in detected_patents_data[:10]:

    print(
        f"Imagen: {patent_info['path']}"
    )

    print(
        f"OCR: {patent_info['patent_ocr']}"
    )

    print("-" * 50)


# Cargar dataset de multas
interim_fines_path: str = (
    "urban_flow/data/interim/speeding_fines.csv"
)

speeding_fines_df: pd.DataFrame = pd.read_csv(
    interim_fines_path
)


# Inicializar nuevas columnas
speeding_fines_df['imagen'] = None
speeding_fines_df['patente_imagen'] = ""
speeding_fines_df['ratio'] = 0.0


print("Relacionando multas con imágenes...")


# Relacionar multas con imágenes OCR
for index, row in speeding_fines_df.iterrows():

    # Limpiar patente original
    original_patent = re.sub(
        r'[^A-Z0-9]',
        '',
        str(row['patente']).upper()
    )

    best_match_ratio = 0.0
    best_matched_image_path = None
    best_matched_patent_ocr = ""

    for detected_img_info in detected_patents_data:

        # Limpiar patente detectada
        current_detected_patent = re.sub(
            r'[^A-Z0-9]',
            '',
            detected_img_info['patent_ocr'].upper()
        )

        # Calcular ratio
        ratio = calculate_match_ratio(
            original_patent,
            current_detected_patent
        )

        # Guardar mejor coincidencia
        if ratio > best_match_ratio:

            best_match_ratio = ratio

            best_matched_image_path = (
                detected_img_info['path']
            )

            best_matched_patent_ocr = (
                current_detected_patent
            )

    # Considerar coincidencia válida si supera el 80%
    if best_match_ratio >= 0.80:

        speeding_fines_df.at[
            index,
            'imagen'
        ] = best_matched_image_path

        speeding_fines_df.at[
            index,
            'patente_imagen'
        ] = best_matched_patent_ocr

        speeding_fines_df.at[
            index,
            'ratio'
        ] = round(best_match_ratio, 2)


print("Relación de multas e imágenes completada.")


# Guardar dataset final
output_processed_fines_path: str = (
    "urban_flow/data/processed/speeding_fines_image.csv"
)

os.makedirs(
    os.path.dirname(output_processed_fines_path),
    exist_ok=True
)

speeding_fines_df.to_csv(
    output_processed_fines_path,
    index=False
)


print(
    f"Dataset final guardado en "
    f"{output_processed_fines_path}"
)


# Mostrar primeras filas para verificación
print("\nDataFrame resultante (primeras 5 filas):")

display(speeding_fines_df.head())
````

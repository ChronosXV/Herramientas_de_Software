`IMPORTANTE: Esta es una aproximación base y sin complicaciones.` Se aceptan otras exploraciones siempre y cuando los resultados sean aceptables. Una base un poco más compleja puede ser:

````python
#@title Instalación de [easyocr](https://github.com/JaidedAI/EasyOCR)
!pip install -q easyocr
````

````python
#@title Definición de la función extraer_patente

import easyocr
def extraer_patente(image_path) -> str:
  """ Extrae la patente de una imagen.

  Args:
      image_path (str): Ruta de la imagen.

  Returns:
      str: Patente extraída.
  """
  reader = easyocr.Reader(['es'])
  detection:list = reader.readtext(image_path)
  if len(detection) == 0:
    return ''
  patent = ''
  for d in detection:
    if isinstance(d,tuple) and len(d) == 3:
      if isinstance(d[1],str):
        patent+=d[1]
  return patent
````

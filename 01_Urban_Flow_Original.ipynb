````python
# Ejercicio 3.2
# Normalizar las horas al formato de 24hs, las horas inválidas se procesaran
# como 00:00. Mostrar las 10 primeras horas.


def parse_mixed_time(time_str: str) -> pd.Timestamp | None:
    """Parsea una cadena de tiempo que puede venir en varios formatos.

    Args:
        time_str (str): La cadena de tiempo a parsear.

    Returns:
        pd.Timestamp | None: Un objeto Timestamp si el parseo es exitoso,
                              None en caso contrario.
    """
    if pd.isna(time_str) or time_str == "invalid_hour":
        return None  # Retorna None para que fillna pueda actuar
    try:
        # Intenta parsear formato HH:MM AM/PM (e.g., '09:01 AM')
        return pd.to_datetime(time_str, format="%I:%M %p")
    except ValueError:
        try:
            # Intenta parsear formato HH:MM (e.g., '18:01')
            return pd.to_datetime(time_str, format="%H:%M")
        except ValueError:
            return None  # Si ambos fallan, retorna None


# Aplicar la función de parseo, convertir a 24hs y rellenar NaNs
df_urban_flow["hora"] = (
    df_urban_flow["hora"]
    .apply(parse_mixed_time)
    .dt.strftime("%H:%M")  # Convierte a formato HH:MM (24 horas)
    .fillna("00:00")  # Rellena los valores None (inválidos) con '00:00'
)

# Mostrar las 10 primeras
display(df_urban_flow["hora"].head(10))
````

````python
# Ejercicio 3.3
# Normalizar las ubicaciones. Quitar caracteres especiales y pasar todo a
# mayúsculas. Mostrar las 10 primeras ubicaciones.
# Permitir letras (incluyendo tildes/ñ), números y espacios
df_urban_flow["ubicacion"] = (
    df_urban_flow["ubicacion"]
    .str.replace(r"[^a-zA-ZáéíóúÁÉÍÓÚñÑ0-9\s]", "", regex=True)
    .str.upper()
)

# limpiar TODO y mantener solo lo básico (ASCII):
# df['ubicacion'] = df['ubicacion'].str.normalize('NFKD')\
# .str.encode('ascii', errors='ignore').str.decode('utf-8').str.upper()

display(df_urban_flow["ubicacion"].head(10))
````

````python
# Ejercicio 3.4
# Limpiar y normalizar las patentes, en caso de error completar con pd.NA
# (no disponible/not available). Quitar caracteres especiales y pasar todo a
# mayúsculas. Mostrar las 10 últimas patentes.
# Quitar caracteres especiales (dejamos solo letras y números)
# Pasar a mayúsculas
# Manejar errores/vacíos con pd.NA
df_urban_flow["patente"] = (
    df_urban_flow["patente"]
    .astype(str)
    .str.replace(r"[^a-zA-Z0-9]", "", regex=True)
    .str.upper()
    .replace("", pd.NA)  # Si después de limpiar queda vacío, ponemos NA
)

# En caso de que hubiera valores nulos originales, asegurar que sean pd.NA
df_urban_flow["patente"] = df_urban_flow["patente"].fillna(pd.NA)

# Mostrar las 10 últimas patentes
display(df_urban_flow["patente"].tail(10))
````

````python
# Ejercicio 3.5
# Eliminar las filas que posean valores relevantes para las multas y que se
# encuentren vacios. Mostrar la cantidad de registros eliminados y cuales son
# las columnas que se observaron
# columnas relevantes
columnas_criticas: list[str] = [
    "patente",
    "fecha",
    "hora",
    "ubicacion",
    "velocidad_registrada",
]

# guardado de cantidad inicial y limpieza
filas_antes: int = len(df_urban_flow)
df: pd.DataFrame = df_urban_flow.dropna(subset=columnas_criticas)
filas_eliminadas: int = filas_antes - len(df)

# Se imprime siguiendo la leyenda en la consigna
print(f"Se eliminaron {filas_eliminadas} filas.")
print("Las columnas observadas son:")
for col in columnas_criticas:
    print(f" - {col}")
````

````python
# Ejercicio 3.6
# Detectar y eliminar los outliers. Mostrar la cantidad de registros eliminados
# y cuales son las columnas que se observaron
# columnas numéricas donde buscar outliers
columnas_outliers: list[str] = ["velocidad_registrada"]

filas_antes: int = len(df_urban_flow)

# técnica de IQR para filtrar
for col in columnas_outliers:
    Q1: float = df_urban_flow[col].quantile(0.25)
    Q3: float = df_urban_flow[col].quantile(0.75)
    IQR: float = Q3 - Q1

    # Definimos límites (el estándar es 1.5 veces el IQR)
    limite_inferior: float = Q1 - 1.5 * IQR
    limite_superior: float = Q3 + 1.5 * IQR

    # Filtramos el DataFrame quedándonos solo con lo que está dentro de los
    # límites
    df_urban_flow = df_urban_flow[
        (df_urban_flow[col] >= limite_inferior)
        & (df_urban_flow[col] <= limite_superior)
    ]

# cantidad de columnas eliminadas
filas_eliminadas: int = filas_antes - len(df_urban_flow)

# leyenda pedida en la consigna
print(f"Se eliminaron {filas_eliminadas} filas.")
print("Las columnas observadas son:")
for col in columnas_outliers:
    print(f" - {col}")
````

````python
# Verificación de rangos para el informe
min_reg = df_urban_flow["velocidad_registrada"].min()
max_reg = df_urban_flow["velocidad_registrada"].max()

print("Verificación de rangos en 'velocidad_registrada':")
print(f" - Valor mínimo: {min_reg} km/h")
print(f" - Valor máximo: {max_reg} km/h")
print(
    "\nConclusión: Los valores se encuentran dentro de un rango "
    "de operación vehicular real (30-130 km/h),"
)
print("lo que explica la ausencia de outliers estadísticos.")
````

````python
# Ejercicio 3.7
# Crear la columna exceso_velocidad_real y calcularla. El exceso de velocidad
# real es la diferencia entre velocidad_registrada y la velocidad_maxima.
# Mostrar las primeras 10 patentes y su exceso de velocidad real.

velocidad_registrada_serie: pd.Series = df_urban_flow["velocidad_registrada"]
velocidad_maxima_serie: pd.Series = df_urban_flow["velocidad_maxima"]

df_urban_flow["exceso_velocidad_real"] = (
    velocidad_registrada_serie - velocidad_maxima_serie
)
# se seleccionan solo esas dos columnas
display(df_urban_flow[["patente", "exceso_velocidad_real"]].head(10))
````

````python
# Ejercicio 3.8
# Crear la columna exceso_velocidad y calcularla. El exceso de velocidad es la
# diferencia entre velocidad_registrada y la velocidad_maxima más un 5%
# de la velocidad máxima. Mostrar las primeras 10 patentes y su exceso de
# velocidad.
# Se calcula el exceso con el margen del 5% (margen de tolerancia)
# (Velocidad Registrada - Velocidad Máxima incrementada en un 5%)
velocidad_registrada: pd.Series = df_urban_flow["velocidad_registrada"]
velocidad_maxima: pd.Series = df_urban_flow["velocidad_maxima"]
df_urban_flow["exceso_velocidad"] = velocidad_registrada - (
    velocidad_maxima * 1.05
)

# Mostrar las primeras 10 patentes y su exceso de velocidad
display(df_urban_flow[["patente", "exceso_velocidad"]].head(10))
````

````python
# Ejercicio 3.9
# Eliminar las filas que no cometieron infracciones por el exceso de velocidad.
# Mostrar cuantas filas se eliminaron
# cantidad de filas
filas_antes = len(df_urban_flow)

# Nos quedamos solo con las filas donde el exceso es mayor a 0
# Esto elimina tanto los negativos como los ceros exactos
df_urban_flow = df_urban_flow[df_urban_flow["exceso_velocidad"] > 0]

# calculo de la diferencia
filas_eliminadas = filas_antes - len(df_urban_flow)

# Se muestra el resultado con la leyenda solicitada en la consigna
print(f"Se eliminaron {filas_eliminadas} filas.")
````

## Ejercicio 04

> Puntos: 2

Definir una clase FineAnalyzer que cumpla con las siguientes funcionalidades:

- Reciba el DataFrame limpio como argumento de inicialización.

- Encapsule correctamente los datos.

- Implemente métodos para:

  - Ranking de las patentes más multadas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de patentes y cantidad..

  - Ranking de los horarios donde se producen más multas (top 5). Debe retornar un DataFrame ordenado de mayor a menor y el índice debe comenzar desde 1, con las columnas de hora y cantidad.


  - Exceso promedio de velocidad. Debe retorna un flotante.

  - Exceso real promedio de velocidad . Debe retorna un flotante.

  - Contabilizar la cantidad de multas por cada  ubicación. Debe retornar un DataFrame ordenado por ubicación mostrando la ubicación y la cantidad.

- Crear el objecto e invocar cada uno de los métodos en celdas separadas.
  

````python
# Analizamos infracciones a partir del dataframe del ejercicio 3
df_limpio: pd.DataFrame = pd.read_csv(
    "urban_flow/data/interim/speeding_fines.csv"
)
````

````python
class FineAnalyzer:
    """Clase para analizar multas por exceso de velocidad.

    Atributos:
        __df (pd.DataFrame): DataFrame con los datos de las multas.
    """

    def __init__(self, df: pd.DataFrame) -> None:
        """Inicializa la clase FineAnalyzer.

        Args:
            df (pd.DataFrame): DataFrame limpio de multas.
        """
        # Encapsulamiento
        self.__df: pd.DataFrame = df.copy()

    def top_patentes(self) -> pd.DataFrame:
        """Retorna el ranking de las 5 patentes más multadas.

        Returns:
            pd.DataFrame: DataFrame con las columnas 'patente' y 'cantidad',
                          indexado desde 1.
        """
        top_patentes_df: pd.DataFrame = (
            self.__df["patente"].value_counts().head(5).reset_index()
        )
        top_patentes_df.columns = ["patente", "cantidad"]
        top_patentes_df.index = range(1, len(top_patentes_df) + 1)
        return top_patentes_df

    def top_horarios(self) -> pd.DataFrame:
        """Retorna el ranking de los 5 horarios con más multas.

        Returns:
            pd.DataFrame: DataFrame con las columnas 'hora' y 'cantidad',
                          indexado desde 1.
        """
        top_horarios_df: pd.DataFrame = (
            self.__df["hora"].value_counts().head(5).reset_index()
        )
        top_horarios_df.columns = ["hora", "cantidad"]
        top_horarios_df.index = range(1, len(top_horarios_df) + 1)
        return top_horarios_df

    def exceso_promedio(self) -> float:
        """Retorna el exceso de velocidad promedio (con tolerancia).

        Returns:
            float: Exceso de velocidad promedio redondeado a 2 decimales.
        """
        # viene calculado del ejercicio 03
        return round(float(self.__df["exceso_velocidad"].mean()), 2)

    def exceso_real_promedio(self) -> float:
        """Retorna el exceso real promedio de velocidad.

        Returns:
            float: Exceso real promedio de velocidad redondeado a 2 decimales.
        """
        # exceso_velocidad_real ya viene calculado del ejercicio 03
        return round(float(self.__df["exceso_velocidad_real"].mean()), 2)

    def multas_por_ubicacion(self) -> pd.DataFrame:
        """Contabiliza la cantidad de multas por cada ubicación.

        Returns:
            pd.DataFrame: DataFrame con las columnas 'ubicacion' y 'cantidad',
                          ordenado por ubicación.
        """
        ranking: pd.DataFrame = (
            self.__df.groupby("ubicacion").size().reset_index()
        )
        ranking.columns = ["ubicacion", "cantidad"]
        return ranking.sort_values(by="ubicacion").reset_index(drop=True)
````

````python
# Creación del objeto
analyzer = FineAnalyzer(df_urban_flow)
````

````python
# Ranking de patentes
print("Top 5 Patentes más multadas:")
display(analyzer.top_patentes())
````

````python
# Ranking de horarios
print("Top 5 Horarios con más multas:")
display(analyzer.top_horarios())
````

````python
# Promedios de exceso
promedio_legal = analyzer.exceso_promedio()
print(
    f"Exceso de velocidad promedio (considerando 5% de tolerancia): "
    f"{promedio_legal} km/h"
)
````

````python
promedio_real = analyzer.exceso_real_promedio()
print(
    f"El exceso de velocidad real promedio de los infractores es de "
    f"{promedio_real} km/h."
)
````

````python
# Multas por ubicación
print("Cantidad de multas por ubicación:")
display(analyzer.multas_por_ubicacion())
````

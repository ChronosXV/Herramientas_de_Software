# Ejercicio 03

> Puntos: 1

Diseñar el modelo lógico según el archivo data/processed/speeding_fines_image.csv de las siguientes entidades:

- Vehiculo

- Multa

- Radar

- Evidencia

Relaciones que deben ser consideradas a la hora del diseño:

- Un vehículo puede tener muchas multas.

- Una multa puede tener una evidencia (opcional).

- Un radar puede generar muchas multas.

El modelo lógico no está relacionado con la base de datos.

````python
from typing import Optional

# @title Definición de Clases del Modelo

class vehiculo:
    def __init__(self, id_patente: int, patente: Optional[str]):
        self.id_patente = id_patente
        self.patente = patente

class radar:
    def __init__(self, radar_id: int, radar: Optional[str], ubicacion: Optional[str], velocidad_maxima_permitida: Optional[float]):
        self.radar_id = radar_id
        self.radar = radar  # Columna 'radar' de tu tabla
        self.ubicacion = ubicacion
        self.velocidad_maxima_permitida = velocidad_maxima_permitida

class evidencia:
    def __init__(self, evidencia_id: int, multa_id: int, ruta_imagen: Optional[str], patente_detectada_imagen: Optional[str], ratio_confianza: Optional[float]):
        self.evidencia_id = evidencia_id
        self.multa_id = multa_id
        self.ruta_imagen = ruta_imagen
        self.patente_detectada_imagen = patente_detectada_imagen
        self.ratio_confianza = ratio_confianza

class multa:
    def __init__(self, multa_id: int, id_patente: int, radar_id: Optional[int], fecha: Optional[str], hora: Optional[str],
                 velocidad_registrada: Optional[float], exceso_velocidad_real: Optional[float], exceso_velocidad_calculado: Optional[float],
                 estado_multa: Optional[str], vehiculo: vehiculo, radar: Optional[radar] = None, evidencia: Optional[evidencia] = None):
        self.multa_id = multa_id
        self.id_patente = id_patente
        self.radar_id = radar_id
        self.fecha = fecha
        self.hora = hora
        self.velocidad_registrada = velocidad_registrada
        self.exceso_velocidad_real = exceso_velocidad_real
        self.exceso_velocidad_calculado = exceso_velocidad_calculado
        self.estado_multa = estado_multa

        # Guardamos también las relaciones con los objetos completos
        self.vehiculo = vehiculo
        self.radar = radar
        self.evidencia = evidencia
````

# Ejercicio 04

> Puntos: 1

Completar la función `procesar_fila_csv` que recibe como parámetro un `dict` y debe cargar los datos de una fila del archivo `data/processed/speeding_fines_image.csv` y mapearla según las clases definidas en el punto anterior.

````python
#@title Función procesar_fila_csv
def procesar_fila_csv(row: dict) -> multa:
  """ Función que recibe un diccionario y lo mapea a una instancia de la clase
  correspondiente.

  Args:
      row (dict): Fila del archivo `data/processed/speeding_fines_image.csv`

  Returns:
      multa: Instancia de la clase multa
  """

  multa_id_raw = row.get('multa_id')
  multa_id_int = int(multa_id_raw) if multa_id_raw is not None and not pd.isna(multa_id_raw) else 0

  # 1. Instanciar la clase vehiculo
  patente_str_raw = row.get('patente')
  patente_str = str(patente_str_raw).strip() if patente_str_raw is not None and not pd.isna(patente_str_raw) else None

  id_patente_int = abs(hash(patente_str)) % (10**8) if patente_str else 0

  obj_vehiculo = vehiculo(
      id_patente=id_patente_int,
      patente=patente_str
  )

  # 2. Instanciar la clase radar (Opcional, puede no venir en el CSV)
  obj_radar = None
  radar_id_raw = row.get('radar_id')

  if radar_id_raw is not None and not pd.isna(radar_id_raw):
      radar_id_str = str(radar_id_raw).strip()
      if radar_id_str:
          try:
              # Extraemos el número del string (ej: 'R03' -> 3) para el ID numérico de tu tabla
              radar_id_int = int(''.join(filter(str.isdigit, radar_id_str)))
          except ValueError:

              radar_id_int = abs(hash(radar_id_str)) % (10**5)

          ubicacion_str = row.get('ubicacion')
          ubicacion_str = str(ubicacion_str).strip() if ubicacion_str is not None and not pd.isna(ubicacion_str) else None

          vel_max = row.get('velocidad_maxima')
          obj_radar = radar(
              radar_id=radar_id_int,
              radar=radar_id_str,
              ubicacion=ubicacion_str,
              velocidad_maxima_permitida=float(vel_max) if vel_max is not None and not pd.isna(vel_max) else None
          )

  # 3. Instanciar la clase evidencia (Opcional, si el registro tiene imagen)
  obj_evidencia = None
  imagen_ruta_raw = row.get('imagen')

  if imagen_ruta_raw is not None and not pd.isna(imagen_ruta_raw):
      imagen_ruta = str(imagen_ruta_raw).strip()
      if imagen_ruta:
          patente_detectada_raw = row.get('patente_imagen')
          patente_detectada_imagen = str(patente_detectada_raw).strip() if patente_detectada_raw is not None and not pd.isna(patente_detectada_raw) else None

          ratio_raw = row.get('ratio')
          ratio_confianza = float(ratio_raw) if ratio_raw is not None and not pd.isna(ratio_raw) else None

          obj_evidencia = evidencia(
              evidencia_id=multa_id_int,
              multa_id=multa_id_int,
              ruta_imagen=imagen_ruta,
              patente_detectada_imagen=patente_detectada_imagen,
              ratio_confianza=ratio_confianza
          )

  # 4. Parsear los datos numéricos de la multa
  vel_reg_raw = row.get('velocidad_registrada')
  velocidad_registrada = float(vel_reg_raw) if vel_reg_raw is not None and not pd.isna(vel_reg_raw) else None

  exc_real_raw = row.get('exceso_velocidad_real')
  exceso_velocidad_real = float(exc_real_raw) if exc_real_raw is not None and not pd.isna(exc_real_raw) else None

  exc_calc_raw = row.get('exceso_velocidad')
  exceso_velocidad_calculado = float(exc_calc_raw) if exc_calc_raw is not None and not pd.isna(exc_calc_raw) else None

  # 5. Instanciar la clase multa relacionando todo
  fecha_str_raw = row.get('fecha')
  fecha_str = str(fecha_str_raw).strip() if fecha_str_raw is not None and not pd.isna(fecha_str_raw) else None

  hora_str_raw = row.get('hora')
  hora_str = str(hora_str_raw).strip() if hora_str_raw is not None and not pd.isna(hora_str_raw) else None

  estado_multa_str_raw = row.get('estado_multa')
  estado_multa_str = str(estado_multa_str_raw).strip() if estado_multa_str_raw is not None and not pd.isna(estado_multa_str_raw) else None

  nueva_multa = multa(
      multa_id=multa_id_int,
      id_patente=obj_vehiculo.id_patente,
      radar_id=obj_radar.radar_id if obj_radar else None,
      fecha=fecha_str,
      hora=hora_str,
      velocidad_registrada=velocidad_registrada,
      exceso_velocidad_real=exceso_velocidad_real,
      exceso_velocidad_calculado=exceso_velocidad_calculado,
      estado_multa=estado_multa_str,
      vehiculo=obj_vehiculo,
      radar=obj_radar,
      evidencia=obj_evidencia
  )

  return nueva_multa
````

# Modelo entidad relacion


  Tabla: Vehiculo

    CREATE TABLE vehiculo (
        id_patente INT NOT NULL,
        patente VARCHAR  NULL,
        PRIMARY KEY (id_patente)
    );

   Tabla: Radar

    CREATE TABLE radar (
        radar_id INT NOT NULL,
        radar VARCHAR  NULL,
        ubicacion VARCHAR NULL,
        velocidad_maxima_permitida FLOAT,
        PRIMARY KEY (radar_id)
    );

  Tabla: Multa
  
    CREATE TABLE multa (
        multa_id INT NOT NULL,
        patente VARCHAR NOT NULL,
        radar_id VARCHAR NULL,
        fecha DATE NOT NULL,
        hora TIME NOT NULL,
        velocidad_registrada DECIMAL(5,2),
        exceso_velocidad_real DECIMAL(5,2),
        exceso_velocidad_calculado DECIMAL(5,2),
        estado_multa VARCHAR,
        PRIMARY KEY (multa_id),
        FOREIGN KEY (id_patente) REFERENCES vehiculo(patente),
        FOREIGN KEY (radar_id) REFERENCES radar(radar_id)
    );

  Tabla: Evidencia
  
    CREATE TABLE evidencia (
        evidencia_id INT NOT NULL,
        ruta_imagen VARCHAR NOT NULL,
        patente_detectada_imagen VARCHAR,
        ratio_confianza DECIMAL(3,2),
        PRIMARY KEY (evidencia_id),
        FOREIGN KEY (multa_id) REFERENCES multa(multa_id)
    );

````python
# @title Definición del Modelo Relacional (SQLAlchemy)

Base = declarative_base()

class VehiculoModel(Base):
    __tablename__ = 'vehiculo'

    id_patente: int = Column(Integer, primary_key=True)
    patente: str = Column(String, nullable=True)

    # Relación: Un vehículo puede tener muchas multas
    multas = relationship("MultaModel", back_populates="vehiculo", cascade="all, delete-orphan")

    def __repr__(self) -> str:
        return f"<Vehiculo(id_patente={self.id_patente}, patente='{self.patente}')>"


class RadarModel(Base):
    __tablename__ = 'radar'

    radar_id: int = Column(Integer, primary_key=True)
    radar: str = Column(String, nullable=True)
    ubicacion: str = Column(String, nullable=True)
    velocidad_maxima_permitida: float = Column(Float, nullable=True)

    # Relación: Un radar puede generar muchas multas
    multas = relationship("MultaModel", back_populates="radar")

    def __repr__(self) -> str:
        return f"<Radar(radar_id={self.radar_id}, ubicacion='{self.ubicacion}')>"


class MultaModel(Base):
    __tablename__ = 'multa'

    multa_id: int = Column(Integer, primary_key=True)
    id_patente: int = Column(Integer, ForeignKey('vehiculo.id_patente'), nullable=False)
    radar_id: int = Column(Integer, ForeignKey('radar.radar_id'), nullable=True)
    fecha = Column(Date, nullable=False)
    hora = Column(Time, nullable=False)
    velocidad_registrada: float = Column( Numeric(5, 2),nullable=True)
    exceso_velocidad_real: float = Column(Numeric(5, 2), nullable=True)
    exceso_velocidad_calculado: float = Column(Numeric(5, 2), nullable=True)
    estado_multa: str = Column(String, nullable=True)

    # Relaciones estipuladas
    vehiculo = relationship("VehiculoModel", back_populates="multas")
    radar = relationship("RadarModel", back_populates="multas")
    # Una multa puede tener una evidencia (opcional) -> uselist=False hace la relación 1 a 1
    evidencia = relationship("EvidenciaModel", uselist=False, back_populates="multa", cascade="all, delete-orphan")

    def __repr__(self) -> str:
        return f"<Multa(multa_id={self.multa_id}, patente_id={self.id_patente}, fecha={self.fecha})>"


class EvidenciaModel(Base):
    __tablename__ = 'evidencia'

    evidencia_id: int = Column(Integer, primary_key=True)
    multa_id: int = Column(Integer, ForeignKey('multa.multa_id'), nullable=False)
    ruta_imagen: str = Column(String, nullable=False)
    patente_detectada_imagen: str = Column(String, nullable=True)
    ratio_confianza: float = Column(Numeric(3, 2), nullable=True)

    # Relación inversa hacia multa
    multa = relationship("MultaModel", back_populates="evidencia")

    def __repr__(self) -> str:
        return f"<Evidencia(evidencia_id={self.evidencia_id}, multa_id={self.multa_id}, ratio={self.ratio_confianza})>"
````

Migrar los datos desde los CSV.

````python
# @title Creación y Población de la Base de Datos transito

# 1. Crear la base de datos (usamos SQLite local por comodidad en Colab)
csv_file_path = os.path.join(os.getcwd(), 'urban_flow/data/processed/speeding_fines_image.csv')

database_url: str = "sqlite:///transito.db"
engine = create_engine(database_url, echo=False)

# 2. Crear las tablas automáticamente
Base.metadata.create_all(engine)
print("📌 Tablas relacionales creadas con éxito en la base de datos.")

# 3. Vincular sesión
Session = sessionmaker(bind=engine)
session = Session()

# 4. Migrar los datos desde el CSV procesado usando la lógica del Ejercicio 04
try:
    # Reutilizamos el path definido o el del ejercicio 4
    df_multas = pd.read_csv(csv_file_path)

    vehiculos_cargados = set()
    radares_cargados = set()

    print("⏳ Iniciando la migración de datos a SQLAlchemy...")

    for _, row_data in df_multas.iterrows():
        row_dict = row_data.to_dict()

        # Mapeamos la fila utilizando tu objeto lógico del Ejercicio 04
        obj_logico = procesar_fila_csv(row_dict)

        # Insertar Vehículo si no se procesó antes
        if obj_logico.vehiculo.id_patente not in vehiculos_cargados:
            db_vehiculo = VehiculoModel(
                id_patente=obj_logico.vehiculo.id_patente,
                patente=obj_logico.vehiculo.patente
            )
            session.add(db_vehiculo)
            vehiculos_cargados.add(obj_logico.vehiculo.id_patente)

        # Insertar Radar si existe y no se procesó antes
        if obj_logico.radar and (obj_logico.radar.radar_id not in radares_cargados):
            db_radar = RadarModel(
                radar_id=obj_logico.radar.radar_id,
                radar=obj_logico.radar.radar,
                ubicacion=obj_logico.radar.ubicacion,
                velocidad_maxima_permitida=obj_logico.radar.velocidad_maxima_permitida
            )
            session.add(db_radar)
            radares_cargados.add(obj_logico.radar.radar_id)

        # Parsear tipos nativos para fechas y tiempos
        fecha_obj = datetime.strptime(obj_logico.fecha, "%Y-%m-%d").date() if obj_logico.fecha else None
        hora_obj = datetime.strptime(obj_logico.hora, "%H:%M").time() if obj_logico.hora else None

        # Insertar Multa
        db_multa = MultaModel(
            multa_id=obj_logico.multa_id,
            id_patente=obj_logico.id_patente,
            radar_id=obj_logico.radar_id if obj_logico.radar_id != 0 else None,
            fecha=fecha_obj,
            hora=hora_obj,
            velocidad_registrada=obj_logico.velocidad_registrada,
            exceso_velocidad_real=obj_logico.exceso_velocidad_real,
            exceso_velocidad_calculado=obj_logico.exceso_velocidad_calculado,
            estado_multa=obj_logico.estado_multa
        )
        session.add(db_multa)

        # Insertar Evidencia (si aplica)
        if obj_logico.evidencia:
            db_evidencia = EvidenciaModel(
                evidencia_id=obj_logico.evidencia.evidencia_id,
                multa_id=obj_logico.evidencia.multa_id,
                ruta_imagen=obj_logico.evidencia.ruta_imagen,
                patente_detectada_imagen=obj_logico.evidencia.patente_detectada_imagen,
                ratio_confianza=obj_logico.evidencia.ratio_confianza
            )
            session.add(db_evidencia)

    session.commit()
    print("✅ Migración completada exitosamente.")

except Exception as e:
    session.rollback()
    print(f"❌ Error durante la migración: {e}")

# 5. Validar cantidad de registros insertados
print("\n--- Validación de Registros en Base de Datos ---")
print(f"Total Vehículos: {session.query(VehiculoModel).count()}")
print(f"Total Radares: {session.query(RadarModel).count()}")
print(f"Total Multas: {session.query(MultaModel).count()}")
print(f"Total Evidencias: {session.query(EvidenciaModel).count()}")
````

En este ejercicio vamos a vincular las 2 bases de datos, la relacional y la vectorial.

Crear la Base De Datos Vectorial `patente_vectorial` para almacenar los datos utilizando el modelo base OpenClip. Ambas instalaciones se realizaron al inicio del notebook.

Para realizar el almacenamiento vamos a tomar el id del Vehículo y el vector de la imagen.

La idea es que al obtener una imagen podamos obtener por aproximación el id del vehiculo.

Poblar la Base De Datos.



````python
# Creamos la query de SQLAlchemy que vamos a utlizar para llegar la tabla "patente_vectorial"

query: Query = session.query(
    MultaModel.id_patente.label("id_vehiculo"),
    EvidenciaModel.ruta_imagen.label("path_imagen")
).join(EvidenciaModel, MultaModel.multa_id == EvidenciaModel.multa_id)

# La cargamos directamente en un DataFrame de Pandas
df_imagenes: pd.DataFrame = pd.read_sql(query.statement, session.bind)

# Mostramos las primeras filas
print(df_imagenes.head())

print("\n--- Visualizando Imágenes ---")
from IPython.display import Image as DisplayImage
import os

for index, row in df_imagenes.iterrows():
    image_path: str = row['path_imagen']
    if os.path.exists(image_path):
        print(f"ID Vehículo: {row['id_vehiculo']}, Ruta: {image_path}")
        display(DisplayImage(filename=image_path, width=200))
    else:
        print(f"¡Advertencia! Imagen no encontrada: {image_path}")
    if index >= 4: # Para no mostrar demasiadas imágenes, limitamos a las primeras 5
        break
````

````python
# Creacion de la base vectorial
# Definimos la ruta de la carpeta donde se guardará la base de datos vectorial

db_path: str = os.path.join(os.getcwd(), "patente_vectorial_db")

# Inicializamos el cliente de ChromaDB indicándole que sea persistente en disco
chroma_client: chromadb.PersistentClient = chromadb.PersistentClient(path=db_path)

# Creamos o recuperamos la colección llamada "patente_vectorial"
# Si ya existía de una ejecución previa, la reutiliza; si no, la crea de cero.
coleccion_vectores: Collection = chroma_client.get_or_create_collection(name="patente_vectorial")

print(f"📁 Base de datos vectorial 'patente_vectorial' inicializada en: {db_path}")

# Configuracion de Openclip
# Detectar si hay GPU (cuda) disponible en Colab para ir más rápido, sino usa CPU
device: str = "cuda" if torch.cuda.is_available() else "cpu"

# Cargamos el modelo OpenClip base (ViT-B-32) y las transformaciones predeterminadas para las imágenes

model: torch.nn.Module
preprocess: Callable
model, _, preprocess = open_clip.create_model_and_transforms(
    'ViT-B-32',
    pretrained='laion2b_s34b_b79k'
)
model = model.to(device)
model.eval()  # Lo ponemos en modo evaluación (apaga dropout/batchnorm para inferencia)

print(f"🚀 Modelo OpenClip (ViT-B-32) cargado exitosamente en [{device}].")
````

````python
# Cargamos la base con los vectores
print(f"⏳ Iniciando el llenado de 'patente_vectorial' con {len(df_imagenes)} imágenes...")

# Contadores para ver cómo va el proceso
exitosos = 0
errores = 0

# Recorremos tu DataFrame fila por fila
for index, row in df_imagenes.iterrows():
    id_vehiculo = str(row['id_vehiculo'])
    image_path = row['path_imagen']

    # 1. Validar que el archivo físico de la imagen exista en Colab
    if not os.path.exists(image_path):
        print(f"⚠️ Archivo omitido (no existe en el disco): {image_path}")
        errores += 1
        continue

    try:
        # 2. Cargar la imagen y prepararla para el modelo
        img = Image.open(image_path).convert("RGB")
        img_preprocesada = preprocess(img).unsqueeze(0).to(device)

        # 3. Generar el embedding (vector de 512 dimensiones) con OpenClip
        with torch.no_grad():
            img_features = model.encode_image(img_preprocesada)
            img_features /= img_features.norm(dim=-1, keepdim=True)
        # Convertimos el resultado de PyTorch a una lista nativa de Python (floats)
        vector_valores = img_features.cpu().numpy().flatten().tolist()

        # 4. Insertar el registro en la colección de ChromaDB
        coleccion_vectores.add(
            ids=[f"id_{index}"],
            embeddings=[vector_valores],
            metadatas=[{
                "id_vehiculo": id_vehiculo,
                "ruta_imagen": image_path
            }]
        )
        exitosos += 1

# Imprimir progreso cada 5 imágenes para no saturar la pantalla
        if exitosos % 5 == 0:
            print(f"🔄 Vectores indexados con éxito: {exitosos}/{len(df_imagenes)}")

    except Exception as e:
        print(f"❌ Error al procesar la imagen de la patente {id_vehiculo} en la ruta {image_path}: {e}")
        errores += 1

# =====================================================================
# 5. VERIFICACIÓN FINAL
# =====================================================================
print("\n--- 🏁 Proceso de Carga Finalizado ---")
print(f"✅ Vectores guardados correctamente: {exitosos}")
print(f"❌ Registros fallidos/omitidos: {errores}")
print(f"📊 Total actual de registros en 'patente_vectorial': {coleccion_vectores.count()}")
````

# Ejercicio 09

> Puntos: 1

Crear la función `buscar_patente_imagen` que recibe como parámetro una imagen y retorna todos los datos del vehículo.

````python
# Creacion de la funcion "buscar_patente_imagen"

def buscar_patente_imagen(ruta_imagen_test: str, top_k: int = 1) -> Optional[Dict[str, Union[int, str, float]]]:
    """
    Recibe la ruta de una imagen (absoluta o relativa), busca su vector más cercano
    en ChromaDB y devuelve los datos del vehículo asociado desde SQLite.
    """
    import os
    import torch
    from PIL import Image

    # --- Ajuste inteligente de rutas para Colab/Entorno local ---
    ruta_real: str = ruta_imagen_test

    # Si no encuentra la ruta original, intentamos normalizarla
    if not os.path.exists(ruta_real):
        if ruta_real.startswith('/content/'):
            # Quita '/content/' y prueba con la ruta relativa (urban_flow/...)
            ruta_real = ruta_real.replace('/content/', '', 1)
        else:
            # Si le pasaron la relativa y Colab exige absoluta, le agrega /content/
            ruta_real = os.path.join('/content', ruta_real)

    # Si después del ajuste sigue sin existir el archivo físico, te avisa cuál es la ruta real que buscó
    if not os.path.exists(ruta_real):
        print(f"❌ Error: El archivo físico no existe en la ruta original ni en la alternativa: {ruta_real}")
        return None

    try:
        # 1. Cargar la imagen usando la ruta real que sí existe en el disco
        img: Image.Image = Image.open(ruta_real).convert("RGB")
        img_preprocesada: torch.Tensor = preprocess(img).unsqueeze(0).to(device)

        # 2. Generar el embedding de la imagen de prueba con OpenClip
        with torch.no_grad():
            img_features: torch.Tensor = model.encode_image(img_preprocesada)
            img_features /= img_features.norm(dim=-1, keepdim=True)

        vector_test: List[float] = img_features.cpu().numpy().flatten().tolist()

        # 3. Buscar en la Base de Datos Vectorial (ChromaDB)
        resultados_vectoriales: Dict[str, Any] = coleccion_vectores.query(
            query_embeddings=[vector_test],
            n_results=top_k
        )

        if not resultados_vectoriales["metadatas"] or len(resultados_vectoriales["metadatas"][0]) == 0:
            print("⚠️ No se encontraron patentes similares en la base vectorial.")
            return None

        # Extraemos los metadatos del mejor resultado
        mejor_coincidencia: Dict[str, Union[int, str]] = resultados_vectoriales["metadatas"][0][0]
        id_vehiculo_detectado: Union[int, str] = mejor_coincidencia["id_vehiculo"]
        distancia: float = resultados_vectoriales["distances"][0][0]

        print(f"🎯 Buscando en ChromaDB... Encontrado ID Vehículo: {id_vehiculo_detectado} (Distancia: {distancia:.4f})")

        # 4. Ir a buscar TODOS los datos del vehículo a la base SQLite con SQLAlchemy
        vehiculo_db: Optional[VehiculoModel] = session.query(VehiculoModel).filter(VehiculoModel.id_patente == id_vehiculo_detectado).first()

        if vehiculo_db:
            datos_vehiculo: Dict[str, Union[int, str, float]] = {
                "id_vehiculo": vehiculo_db.id_patente,
                "patente_texto": vehiculo_db.patente,
                "ruta_imagen_origen": mejor_coincidencia["ruta_imagen"],
                "score_similitud": round(float(1 - distancia), 4)
            }
            return datos_vehiculo
        else:
            print(f"⚠️ El ID {id_vehiculo_detectado} se halló en ChromaDB pero no existe en SQLite.")
            return None

    except Exception as e:
        print(f"❌ Error durante la búsqueda: {e}")
        return None
````
